# Quarterly Revenue Analysis

This notebook compares 2026 revenue with 2025 and fits a simple growth model
$$
R_q = R_0 \, (1 + g)^q
$$
where $g$ is the quarterly growth rate.

```python
import pandas as pd
revenue = pd.DataFrame({'2026': [1.42, 1.58, 1.71, 1.96], '2025': [1.21, 1.30, 1.44, 1.52]}, index=['Q1','Q2','Q3','Q4'])
revenue
```

## Growth

The chart shows both years side by side.

```python
revenue.plot.bar()
```

```python
from sympy import symbols, Eq
R0, g, q = symbols('R_0 g q')
Eq(symbols('R_q'), R0*(1+g)**q)
```

```python
growth = (revenue['2026'] / revenue['2025'] - 1).mean()
print(f'Average growth: {growth:.1%}')
```
